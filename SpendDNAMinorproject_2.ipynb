FEATURE 1: Transaction Parser

```python
# AI-assisted: Syntax assistance for cleaning mixed date and amount formats.

df_clean = df.copy()

# -----------------------------
# DATE CLEANING
# -----------------------------
df_clean["date"] = pd.to_datetime(
    df_clean["Date"],
    errors="coerce",
    dayfirst=True
)

# -----------------------------
# AMOUNT CLEANING
# -----------------------------
df_clean["amount"] = (
    df_clean["Amount"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Rs.", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_clean["amount"] = pd.to_numeric(
    df_clean["amount"],
    errors="coerce"
)

# -----------------------------
# TYPE STANDARDIZATION
# -----------------------------
df_clean["type"] = (
    df_clean["Type"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df_clean["type"] = df_clean["type"].replace({
    "dr": "debit",
    "debit": "debit",
    "cr": "credit",
    "credit": "credit"
})

# -----------------------------
# MODE CLEANING
# -----------------------------
df_clean["Mode"] = (
    df_clean["Mode"]
    .replace("", np.nan)
)

# -----------------------------
# DUPLICATES
# -----------------------------
duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates().copy()

# -----------------------------
# INVALID DATA CHECK
# -----------------------------
invalid_dates = df_clean["date"].isna().sum()
invalid_amounts = df_clean["amount"].isna().sum()

print("=" * 70)
print("FEATURE 1 - TRANSACTION PARSER")
print("=" * 70)

print("Original transactions :", len(df))
print("Duplicate rows removed:", duplicates_before)
print("Clean transactions    :", len(df_clean))
print("Unparseable dates     :", invalid_dates)
print("Unparseable amounts   :", invalid_amounts)

print("\nCleaned data types:")
print(df_clean[["date", "amount", "type"]].dtypes)
```

Vendor Extraction Function

```python
def extract_vendor(description):

    description = str(description).upper()

    # Special case: ATM withdrawal
    if "ATM-WDL" in description:
        return "Cash Withdrawal"

    # Special case: rent
    if "RENT-LANDLORD" in description:
        return "Rent"

    # Special case: salary
    if "SALARY" in description:
        return "Salary"

    # Check canonical vendor keywords
    for vendor, keywords in vendor_keywords.items():

        for keyword in keywords:

            if keyword.upper() in description:
                return vendor

    # P2P transfers
    if "UPI-" in description and "@" in description:
        return "P2P Transfer"

    return "Uncategorised"


df_clean["vendor_clean"] = df_clean["Description"].apply(extract_vendor)

print("=" * 70)
print("FEATURE 2 - VENDOR EXTRACTOR")
print("=" * 70)

print("Unique vendors:", df_clean["vendor_clean"].nunique())

print("\nTop vendors:")
print(df_clean["vendor_clean"].value_counts().head(15))
```

FEATURE 7: Z-Score Anomaly Detection

```python
# AI-assisted: Implemented the category-wise z-score formula specified
# in the Unlox Academy project brief.

analysis_debits = debits[
    debits["category"].isin(consumption_categories)
].copy()

category_mean = (
    analysis_debits
    .groupby("category")["amount"]
    .transform("mean")
)

category_std = (
    analysis_debits
    .groupby("category")["amount"]
    .transform("std")
)

analysis_debits["z_score"] = (
    analysis_debits["amount"] - category_mean
) / category_std

anomalies = analysis_debits[
    analysis_debits["z_score"] > 2
].copy()

anomalies = anomalies.sort_values(
    "z_score",
    ascending=False
)

print("=" * 70)
print("FEATURE 7 - ANOMALY DETECTION")
print("=" * 70)

print("Anomalous transactions:", len(anomalies))
```

FEATURE 8: Archetype Functions

```python
# AI-assisted: Archetype functions follow the quantitative rules
# defined in the Unlox Academy project brief.

debit_total = debits["amount"].sum()


def percentage_of_debits(category):

    amount = debits[
        debits["category"] == category
    ]["amount"].sum()

    if debit_total == 0:
        return 0

    return (amount / debit_total) * 100


def detect_foodie():

    food_categories = [
        "Food Delivery",
        "Restaurants",
        "Cafe"
    ]

    amount = debits[
        debits["category"].isin(food_categories)
    ]["amount"].sum()

    percentage = (
        amount / debit_total
    ) * 100

    return percentage > 25, percentage


def detect_quick_commerce():

    percentage = percentage_of_debits(
        "Quick Commerce"
    )

    return percentage > 15, percentage


def detect_shopaholic():

    percentage = percentage_of_debits(
        "E-commerce"
    )

    return percentage > 15, percentage


def detect_investor():

    percentage = percentage_of_debits(
        "Investments"
    )

    return percentage > 15, percentage


def detect_late_night_snacker():

    food = debits[
        debits["category"] == "Food Delivery"
    ]

    if len(food) == 0:
        return False, 0

    late_night = food[
        (food["hour"] >= 21) |
        (food["hour"] <= 2)
    ]

    percentage = (
        len(late_night) / len(food)
    ) * 100

    return percentage > 50, percentage


def detect_cab_commuter():

    percentage = percentage_of_debits(
        "Transport"
    )

    return percentage > 10, percentage


def detect_subscription_lover():

    subscription_vendors = debits[
        debits["category"] == "Subscriptions"
    ]["vendor_clean"].nunique()

    return subscription_vendors >= 5, subscription_vendors


def detect_yolo_spender():

    return savings_rate < 10, savings_rate


def detect_disciplined_saver():

    return savings_rate > 40, savings_rate
```

Run All Archetypes

```python
print("=" * 70)
print("RAHUL'S SPENDING ARCHETYPES")
print("=" * 70)

archetypes = []

foodie_flag, foodie_value = detect_foodie()

if foodie_flag:
    archetypes.append(
        ("THE FOODIE", foodie_value, "% on Food + Restaurant + Cafe")
    )

quick_flag, quick_value = detect_quick_commerce()

if quick_flag:
    archetypes.append(
        ("THE QUICK COMMERCE JUNKIE", quick_value, "% on Quick Commerce")
    )

shop_flag, shop_value = detect_shopaholic()

if shop_flag:
    archetypes.append(
        ("THE SHOPAHOLIC", shop_value, "% on E-commerce")
    )

investor_flag, investor_value = detect_investor()

if investor_flag:
    archetypes.append(
        ("THE INVESTOR", investor_value, "% on Investments")
    )

late_flag, late_value = detect_late_night_snacker()

if late_flag:
    archetypes.append(
        ("THE LATE-NIGHT SNACKER", late_value, "% Food Delivery after 9 PM")
    )

cab_flag, cab_value = detect_cab_commuter()

if cab_flag:
    archetypes.append(
        ("THE CAB COMMUTER", cab_value, "% on Transport")
    )

subscription_flag, subscription_value = detect_subscription_lover()

if subscription_flag:
    archetypes.append(
        (
            "THE SUBSCRIPTION LOVER",
            subscription_value,
            "active subscription vendors"
        )
    )

yolo_flag, yolo_value = detect_yolo_spender()

if yolo_flag:
    archetypes.append(
        (
            "THE YOLO SPENDER",
            yolo_value,
            "savings rate"
        )
    )

saver_flag, saver_value = detect_disciplined_saver()

if saver_flag:
    archetypes.append(
        (
            "THE DISCIPLINED SAVER",
            saver_value,
            "savings rate"
        )
    )


for name, value, metric in archetypes:

    if "subscription" in metric.lower():
        print(
            f"-> {name:<30} "
            f"({value:.0f} {metric})"
        )

    else:
        print(
            f"-> {name:<30} "
            f"({value:.2f} {metric})"
        )

print("\nTotal archetypes detected:", len(archetypes))
```

```python
print("=" * 70)
print("BONUS - DAY-OF-WEEK ANALYSIS")
print("=" * 70)

weekday_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_spending = (
    debits
    .groupby("day_of_week")["amount"]
    .sum()
    .reindex(weekday_order)
    .fillna(0)
)

for day, amount in day_spending.items():

    print(
        f"{day:<12} ₹{amount:>12,.2f}"
    )

weekend_spend = day_spending[
    ["Saturday", "Sunday"]
].sum()

weekday_spend = day_spending[
    [
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday"
    ]
].sum()

print("\nWeekend spend :", f"₹{weekend_spend:,.2f}")
print("Weekday spend :", f"₹{weekday_spend:,.2f}")

if weekday_spend > 0:

    weekend_difference = (
        (weekend_spend - weekday_spend)
        / weekday_spend
    ) * 100

    print(
        f"Weekend vs weekday difference: "
        f"{weekend_difference:.2f}%"
    )
```

BONUS: 3-Month Spend Forecast

```python
print("=" * 70)
print("BONUS - NEXT MONTH SPEND FORECAST")
print("=" * 70)

monthly_total = (
    debits
    .groupby("month")["amount"]
    .sum()
    .reindex(range(1, 7))
    .fillna(0)
)

for month, amount in monthly_total.items():

    print(
        f"{month_names[month]} : ₹{amount:,.2f}"
    )

last_three_months = monthly_total.iloc[-3:].values

forecast = np.mean(last_three_months)

print(
    f"\nForecast for July 2024 "
    f"(3-month average): ₹{forecast:,.2f}"
)
```

Final SpendDNA Report

```python
# ============================================================
# FINAL SPENDDNA REPORT
# ============================================================

print("\n" + "=" * 75)
print("                    SPENDDNA FINAL REPORT")
print("                 Your Wallet's Financial DNA")
print("=" * 75)

# ------------------------------------------------------------
# 1. EXECUTIVE FINANCIAL SUMMARY
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("1. EXECUTIVE FINANCIAL SUMMARY")
print("-" * 75)

print(f"Total Transactions : {transaction_count:,}")
print(f"Unique Vendors     : {unique_vendors:,}")
print(f"Total Credits      : ₹{total_credits:,.2f}")
print(f"Total Debits       : ₹{total_debits:,.2f}")
print(f"Net Change         : ₹{net_change:,.2f}")
print(f"Savings Rate       : {savings_rate:.2f}%")

# ------------------------------------------------------------
# 2. CATEGORY-WISE SPENDING
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("2. CATEGORY-WISE SPENDING")
print("-" * 75)

if len(category_spend) > 0:

    total_category_spend = category_spend.sum()

    for category, amount in category_spend.sort_values(
        ascending=False
    ).items():

        percentage = (
            amount / total_category_spend * 100
            if total_category_spend > 0 else 0
        )

        bar = "█" * max(1, int(percentage / 2))

        print(
            f"{str(category):<22} "
            f"₹{amount:>12,.2f}  "
            f"{percentage:>6.2f}%  "
            f"{bar}"
        )

# ------------------------------------------------------------
# 3. TOP VENDORS
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("3. TOP VENDORS BY SPENDING")
print("-" * 75)

if len(vendor_spend) > 0:

    for i, (vendor, amount) in enumerate(
        vendor_spend.sort_values(
            ascending=False
        ).head(10).items(),
        start=1
    ):
        print(
            f"{i:>2}. "
            f"{str(vendor):<25} "
            f"₹{amount:>12,.2f}"
        )

# ------------------------------------------------------------
# 4. MONTHLY SPENDING TRENDS
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("4. MONTHLY SPENDING TRENDS")
print("-" * 75)

if len(monthly_total) > 0:

    max_month_value = monthly_total.max()

    for month, amount in monthly_total.items():

        bar_length = (
            int((amount / max_month_value) * 35)
            if max_month_value > 0 else 0
        )

        print(
            f"{str(month):<12} "
            f"₹{amount:>12,.2f}  "
            f"{'█' * bar_length}"
        )

# ------------------------------------------------------------
# 5. TIME-OF-DAY PATTERNS
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("5. TIME-OF-DAY SPENDING PATTERNS")
print("-" * 75)

print(
    f"Late-night food delivery : "
    f"{late_night_percentage:.2f}%"
)

print(
    f"Morning cafe spending    : "
    f"{morning_cafe_percentage:.2f}%"
)

if late_night_percentage > 20:
    print("Insight: Significant late-night food spending detected.")
else:
    print("Insight: Late-night food spending is relatively controlled.")

if morning_cafe_percentage > 15:
    print("Insight: Noticeable morning cafe spending detected.")
else:
    print("Insight: Morning cafe spending is relatively controlled.")

# ------------------------------------------------------------
# 6. ANOMALIES
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("6. TOP ANOMALIES")
print("-" * 75)

if len(anomalies) == 0:

    print("No significant anomalies detected.")

else:

    for _, row in anomalies.head(10).iterrows():

        # Prevent NaT strftime error
        if pd.isna(row['date']):
            date_str = "Unknown"
        else:
            date_str = row['date'].strftime('%d %b %Y')

        vendor = str(row.get('vendor_clean', 'Unknown'))
        category = str(row.get('category', 'Unknown'))
        amount = row.get('amount', 0)

        print(
            f"{date_str:<15} | "
            f"{vendor:<22} | "
            f"{category:<18} | "
            f"₹{amount:>10,.2f}"
        )

# ------------------------------------------------------------
# 7. SPENDING ARCHETYPE
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("7. SPENDING ARCHETYPE")
print("-" * 75)

if isinstance(archetypes, (list, tuple)):

    for archetype in archetypes:
        print(f"• {archetype}")

else:

    print(archetypes)

# ------------------------------------------------------------
# 8. KEY INSIGHTS
# ------------------------------------------------------------

print("\n" + "-" * 75)
print("8. KEY INSIGHTS")
print("-" * 75)

# Highest spending category
if len(category_spend) > 0:

    top_category = category_spend.idxmax()
    top_category_amount = category_spend.max()

    print(
        f"• Highest spending category: "
        f"{top_category} "
        f"(₹{top_category_amount:,.2f})"
    )

# Highest spending vendor
if len(vendor_spend) > 0:

    top_vendor = vendor_spend.idxmax()
    top_vendor_amount = vendor_spend.max()

    print(
        f"• Highest spending vendor: "
        f"{top_vendor} "
        f"(₹{top_vendor_amount:,.2f})"
    )

# Savings insight
if savings_rate >= 20:

    print(
        f"• Savings performance is strong at "
        f"{savings_rate:.2f}%."
    )

elif savings_rate >= 10:

    print(
        f"• Savings performance is moderate at "
        f"{savings_rate:.2f}%."
    )

else:

    print(
        f"• Savings rate is relatively low at "
        f"{savings_rate:.2f}%; spending can be reviewed."
    )

# Anomaly insight
print(
    f"• {len(anomalies)} potentially unusual "
    f"transaction(s) were identified."
)

# ------------------------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("                    FINAL SPENDDNA SUMMARY")
print("=" * 75)

print(
    f"""
SpendDNA analyzed {transaction_count:,} transactions
across {unique_vendors:,} unique vendors.

Total money received : ₹{total_credits:,.2f}
Total money spent    : ₹{total_debits:,.2f}
Net financial change : ₹{net_change:,.2f}
Savings rate         : {savings_rate:.2f}%

The analysis identified the major spending categories,
frequent vendors, monthly spending patterns, time-of-day
behaviour and potentially unusual transactions.

These insights can help the user understand spending habits,
identify unusual expenses and make better financial decisions.
"""
)

print("=" * 75)
print("                    END OF REPORT")
print("=" * 75)
```
